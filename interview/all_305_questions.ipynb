### 🔹 [Q001 of 305] Take a number and print whether it's positive, negative, or zero.
- 📁 **File:** `phase01_conditional_thinking/level01_simple_conditions/Question001.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 01: Simple Conditions` (File Question #1)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The if/else checks test the important cases in order: `number > 0`, `number < 0`.
> - Only the branch whose condition becomes true prints its message.


```python
def main():
    # Question 1: : Take a number and print whether it's positive, negative, or zero.
    number = -7
    if number > 0:
        print("Positive")
    elif number < 0:
        print("Negative")
    else:
        print("Zero")


if __name__ == "__main__":
    main()
```

### 🔹 [Q005 of 305] Check if a given year is a leap year.

> - The condition `(year % 400 == 0) || (year % 4 == 0 && year % 100 != 0)` decides whether the current value matches the requirement.

### 🔹 [Q009 of 305] Take a character and check if it's a vowel or consonant.
- 📁 **File:** `phase01_conditional_thinking/level01_simple_conditions/Question009.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 01: Simple Conditions` (File Question #9)

**💡 Logic & Explanation:**
> - The value stored in `ch` is the input that the conditions or loops work on.
> - The helper checks characters against `a`, `e`, `i`, `o`, and `u`, so the main loop only needs to use that true/false result.
> - The if/else checks test the important cases in order: `Character.isLetter(ch)`, `isVowel(ch)`.
> - Only the branch whose condition becomes true prints its message.


```python
def isVowel(ch):
    lower = ch.lower()
    return lower == 'a' or lower == 'e' or lower == 'i' or lower == 'o' or lower == 'u'

def main():
    # Question 9: : Take a character and check if it's a vowel or consonant.
    ch = 'e'
    if ch.isalpha():
        if isVowel(ch):
            print("Vowel")
        else:
            print("Consonant")
    else:
        print("Not an alphabet")


if __name__ == "__main__":
    main()
```

### 🔹 [Q010 of 305] Take a character and check whether it's uppercase, lowercase, a digit, or a special character.
- 📁 **File:** `phase01_conditional_thinking/level01_simple_conditions/Question010.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 01: Simple Conditions` (File Question #10)

**💡 Logic & Explanation:**
> - The value stored in `ch` is the input that the conditions or loops work on.
> - The if/else checks test the important cases in order: `Character.isUpperCase(ch)`, `Character.isLowerCase(ch)`, `Character.isDigit(ch)`.
> - Only the branch whose condition becomes true prints its message.


```python
def main():
    # Question 10: : Take a character and check whether it's uppercase, lowercase, a digit, or a special character.
    ch = 'A'
    if ch.isupper():
        print("Uppercase")
    elif ch.islower():
        print("Lowercase")
    elif ch.isdigit():
        print("Digit")
    else:
        print("Special character")


if __name__ == "__main__":
    main()
```

### 🔹 [Q012 of 305] If the sides form a valid triangle, determine whether it is equilateral, isosceles, or scalene.
- 📁 **File:** `phase01_conditional_thinking/level02_nested_if_multiple_conditions/Question012.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 02: Nested If Multiple Conditions` (File Question #12)

**💡 Logic & Explanation:**
> - The value stored in `a` is the input that the conditions or loops work on.
> - The if/else checks test the important cases in order: `a + b > c && a + c > b && b + c > a`, `a == b && b == c`, `a == b || b == c || a == c`.
> - Only the branch whose condition becomes true prints its message.


```python
def main():
    # Question 12: : If the sides form a valid triangle, determine whether it is equilateral, isosceles, or scalene.
    a = 5
    b = 5
    c = 8
    if a + b > c and a + c > b and b + c > a:
        if a == b and b == c:
            print("Equilateral")
        elif a == b or b == c or a == c:
            print("Isosceles")
        else:
            print("Scalene")
    else:
        print("Invalid triangle")


if __name__ == "__main__":
    main()
```

> - The if/else checks test the important cases in order: `first != 0 && second % first == 0`, `second != 0 && first % second == 0`.

```python
def main():
    # Question 14: : Check if one of two given numbers is a multiple of the other.
    first = 12
    second = 36
    if first != 0 and second % first == 0:
        print("Second is a multiple of first")
    elif second != 0 and first % second == 0:
        print("First is a multiple of second")
    else:
        print("No number is a multiple of the other")


if __name__ == "__main__":
    main()
```

### 🔹 [Q021 of 305] Take a 3-digit number and check if all digits are distinct.
- 📁 **File:** `phase01_conditional_thinking/level03_math_number_logic/Question021.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 03: Math Number Logic` (File Question #21)

**💡 Logic & Explanation:**
> - The variables `hundreds`, `tens`, and `ones` hold the values that the logic checks, counts, or transforms.
> - The condition `hundreds != tens && tens != ones && hundreds != ones` decides whether the current value matches the requirement.
> - Only the branch whose condition becomes true prints its message.


```python
def main():
    # Question 21: : Take a 3-digit number and check if all digits are distinct.
    number = 427
    hundreds = number // 100
    tens = (number // 10) % 10
    ones = number % 10
    if hundreds != tens and tens != ones and hundreds != ones:
        print("All digits are distinct")
    else:
        print("Digits are not distinct")


if __name__ == "__main__":
    main()
```

### 🔹 [Q022 of 305] Take a 3-digit number and determine if the middle digit is the largest, smallest, or neither.

**💡 Logic & Explanation:**
> - The variables `hundreds`, `middle`, and `ones` hold the values that the logic checks, counts, or transforms.
> - The if/else checks test the important cases in order: `middle > hundreds && middle > ones`, `middle < hundreds && middle < ones`.

### 🔹 [Q030 of 305] Check whether a number is a perfect square (without using the square root function).
- 📁 **File:** `phase01_conditional_thinking/level03_math_number_logic/Question030.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 03: Math Number Logic` (File Question #30)

**💡 Logic & Explanation:**
> - The value stored in `perfectSquare` is the input that the conditions or loops work on.
> - The `for` loop starts with `int i = 1`, keeps running while `i * i <= number` is true, and updates using `i++`.
> - The condition `i * i == number` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 30: : Check whether a number is a perfect square (without using the square root function).
    number = 49
    perfectSquare = False
    i = 1
    while i * i <= number:
        if i * i == number:
            perfectSquare = True
            break
        i += 1
    print("Perfect square" if perfectSquare else "Not a perfect square")


if __name__ == "__main__":
    main()
```

### 🔹 [Q040 of 305] Take a password string and check basic rules (length >= 8 and contains at least one digit).
- 📁 **File:** `phase01_conditional_thinking/level04_logical_operators_compound/Question040.py`
- 🏷️ **Phase:** `Phase 01: Conditional Thinking` | **Level/Category:** `Level 04: Logical Operators Compound` (File Question #40)

**💡 Logic & Explanation:**
> - The variables `password`, `hasDigit`, and `ch` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < password.length()` is true, and updates using `i++`.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The if/else checks test the important cases in order: `Character.isDigit(ch)`, `Character.isUpperCase(ch)`, `Character.isLowerCase(ch)`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 40: : Take a password string and check basic rules (length >= 8 and contains at least one digit).
    password = "Code123not "
    hasDigit = False
    hasUpper = False
    hasLower = False
    hasSpecial = False
    for i in range(0, len(password)):
        ch = password[i]
        if ch.isdigit():
            hasDigit = True
        elif ch.isupper():
            hasUpper = True
        elif ch.islower():
            hasLower = True
        else:
            hasSpecial = True
    if len(password) >= 8 and hasDigit and hasUpper and hasLower and hasSpecial:
        print("Valid password")
    else:
        print("Invalid password")


if __name__ == "__main__":
    main()
```

```python
def sumDigits(number):
    value = abs(number)
    sum = 0
    while True:
        sum += value % 10
        value //= 10
        if not (value > 0):
            break
    return sum

def productDigits(number):
    value = abs(number)
    product = 1
    if value == 0:
        return 0
    while value > 0:
        product *= value % 10
        value //= 10
    return product

def main():
    # Question 48: : Take an integer (1-9999) and check if the sum of its digits is greater than the product of its digits.
    number = 1234
    sum = sumDigits(number)
    product = productDigits(number)
    if sum > product:
        print("Digit sum is greater")
    else:
        print("Digit product is greater or equal")


if __name__ == "__main__":
    main()
```

### 🔹 [Q062 of 305] Print the reverse of a given number.
- 📁 **File:** `phase02_looping_patterns/level02_number_based_looping/Question012.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 02: Number Based Looping` (File Question #12)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper builds the answer from the last digit or last character toward the first one.
> - The calculated answer is printed after the logic produces the final value.


```python
def reverseNumber(number):
    value = abs(number)
    reversed = 0
    while True:
        reversed = reversed * 10 + value % 10
        value //= 10
        if not (value > 0):
            break
    return -reversed if number < 0 else reversed

def main():
    # Question 12: : Print the reverse of a given number.
    number = 12345
    print("Reverse = " + str(reverseNumber(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q063 of 305] Check if a number is a palindrome.
- 📁 **File:** `phase02_looping_patterns/level02_number_based_looping/Question013.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 02: Number Based Looping` (File Question #13)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper compares the original value with its reversed form, because a palindrome reads the same both ways.
> - The calculated answer is printed after the logic produces the final value.


```python
def reverseNumber(number):
    value = abs(number)
    reversed = 0
    while True:
        reversed = reversed * 10 + value % 10
        value //= 10
        if not (value > 0):
            break
    return -reversed if number < 0 else reversed

def isPalindromeNumber(number):
    return abs(number) == abs(reverseNumber(number))

def main():
    # Question 13: : Check if a number is a palindrome.
    number = 12321
    print("Palindrome" if isPalindromeNumber(number) else "Not palindrome")


if __name__ == "__main__":
    main()
```

### 🔹 [Q065 of 305] Check if a number is an Armstrong number.
- 📁 **File:** `phase02_looping_patterns/level02_number_based_looping/Question015.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 02: Number Based Looping` (File Question #15)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper counts how many digits the number has by repeatedly dividing a copy of the number by 10.
> - Then it extracts each digit with `% 10`, raises that digit to the digit-count power, and adds it to `sum`.
> - If the final `sum` equals the original number, the helper returns true.
> - The calculated answer is printed after the logic produces the final value.


```python
def isArmstrong(number):
    value = number
    digits = 0
    while True:
        digits += 1
        value //= 10
        if not (value > 0):
            break
    sum = 0
    value = number
    while True:
        digit = value % 10
        power = 1
        for i in range(1, digits + 1):
            power *= digit
        sum += power
        value //= 10
        if not (value > 0):
            break
    return sum == number

def main():
    # Question 15: : Check if a number is an Armstrong number.
    number = 153
    print("Armstrong number" if isArmstrong(number) else "Not Armstrong number")


if __name__ == "__main__":
    main()
```

### 🔹 [Q066 of 305] Check if a number is a perfect number.
- 📁 **File:** `phase02_looping_patterns/level02_number_based_looping/Question016.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 02: Number Based Looping` (File Question #16)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The calculated answer is printed after the logic produces the final value.


```python
def isPerfect(number):
    if number <= 1:
        return False
    sum = 1
    for factor in range(2, number / 2 + 1):
        if number % factor == 0:
            sum += factor
    return sum == number

def main():
    # Question 16: : Check if a number is a perfect number.
    number = 28
    print("Perfect number" if isPerfect(number) else "Not perfect number")


if __name__ == "__main__":
    main()
```

### 🔹 [Q067 of 305] Print all prime numbers between 1 and 100.
- 📁 **File:** `phase02_looping_patterns/level02_number_based_looping/Question017.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 02: Number Based Looping` (File Question #17)

**💡 Logic & Explanation:**
> - The value stored in `n` is the input that the conditions or loops work on.
> - The helper tries possible divisors from 2 up to the square root of the number.
> - If any divisor divides the number exactly, the number is not prime; otherwise it is prime.
> - The `for` loop starts with `int number = 2`, keeps running while `number <= n` is true, and updates using `number++`.
> - The condition `isPrime(number)` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def isPrime(number):
    if number <= 1:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True

def main():
    # Question 17: : Print all prime numbers between 1 and 100.
    n = 100
    for number in range(2, n + 1):
        if isPrime(number):
            print(number)


if __name__ == "__main__":
    main()
```

**💡 Logic & Explanation:**
> - The variables `first`, and `remainder` hold the values that the logic checks, counts, or transforms.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.
> - The calculated answer is printed after the logic produces the final value.


```python
def main():
    # Question 24: : Find HCF (GCD) of two numbers using loops.
    first = 48
    second = 18
    while second != 0:
        remainder = first % second
        first = second
        second = remainder
    print("GCD = " + str(abs(first)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q075 of 305] Find LCM of two numbers using loops.
- 📁 **File:** `phase02_looping_patterns/level03_mathematical_logical_patterns/Question025.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 03: Mathematical Logical Patterns` (File Question #25)

**💡 Logic & Explanation:**
> - The variables `first`, `a`, `remainder`, and `lcm` hold the values that the logic checks, counts, or transforms.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.
> - The calculated answer is printed after the logic produces the final value.


```python
def main():
    # Question 25: : Find LCM of two numbers using loops.
    first = 12
    second = 18
    a = first
    b = second
    while b != 0:
        remainder = a % b
        a = b
        b = remainder
    lcm = abs(first * second) / abs(a)
    print("LCM = " + str(lcm))


if __name__ == "__main__":
    main()
```

### 🔹 [Q078 of 305] Check if a number is a strong number (sum of factorials of digits = number).
- 📁 **File:** `phase02_looping_patterns/level03_mathematical_logical_patterns/Question028.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 03: Mathematical Logical Patterns` (File Question #28)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper multiplies all numbers from 2 up to the given number to build the factorial step by step.
> - The calculated answer is printed after the logic produces the final value.


```python
def factorial(number):
    result = 1
    for value in range(2, number + 1):
        result *= value
    return result

def main():
    # Question 28: : Check if a number is a strong number (sum of factorials of digits = number).
    number = 5
    print(str(number) + "not  = " + str(factorial(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q090 of 305] Print Stars and Spaces Alternating (Stars and Blank Spaces).
- 📁 **File:** `phase02_looping_patterns/level04_pattern_printing/Question010.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 04: Pattern Printing` (File Question #10)

**💡 Logic & Explanation:**
> - The outer loop controls how many rows are printed.
> - The first inner loop prints the leading blank spaces so the pattern is right-aligned.
> - The second inner loop prints stars, and it prints one blank space after each star except the last star in that row.


```python
def main():
    # Question 10: Print Stars and Spaces Alternating (Stars and Blank Spaces).
    rows = 5

    for row in range(1, rows + 1):
        for space in range(1, rows - row + 1):
            print(" ", end="")

        for star in range(1, row + 1):
            print("*", end="")
            if star < row:
                print(" ", end="")
        print()


if __name__ == "__main__":
    main()
```

### 🔹 [Q101 of 305] Print a centered palindrome number pyramid.
- 📁 **File:** `phase02_looping_patterns/level04_pattern_printing/Question021.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 04: Pattern Printing` (File Question #21)

**💡 Logic & Explanation:**
> - We reverse the number by taking one digit at a time from the end.
> - After the reverse is ready, the code compares it with the original number.
> - If both values are equal, the number reads the same forward and backward, so it is a palindrome.


```python
def main():
    # Question 21: Print a centered palindrome number pyramid.
    rows = 5

    for row in range(1, rows + 1):
        for space in range(1, rows - row + 1):
            print(" ", end="")
        for number in range(1, row + 1):
            print(number, end="")
        for number in range(row - 1, 1 - 1, -1):
            print(number, end="")
        print()


if __name__ == "__main__":
    main()
```

> - Inner loops print the required spaces, stars, numbers, or letters for each row.
> - After each row is complete, System.out.println() moves the output to the next line.


```python
def main():
    # Question 24: Print a centered diamond star pattern.
    rows = 5

    for row in range(1, rows + 1):
        for space in range(1, rows - row + 1):
            print(" ", end="")
        for star in range(1, 2 * row - 1 + 1):
            print("*", end="")
        print()

    for row in range(rows - 1, 1 - 1, -1):
        for space in range(1, rows - row + 1):
            print(" ", end="")
        for star in range(1, 2 * row - 1 + 1):
            print("*", end="")
        print()


if __name__ == "__main__":
    main()
```

> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The calculated answer is printed after the logic produces the final value.


```python
def sumDigits(number):
    value = abs(number)
    sum = 0
    while True:
        sum += value % 10
        value //= 10
        if not (value > 0):
            break
    return sum

def main():
    # Question 31: : Print all numbers whose sum of digits is even (1-100).
    number = 9876
    print("Sum of digits = " + str(sumDigits(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q108 of 305] Print all numbers that are palindromes between 1-500.
- 📁 **File:** `phase02_looping_patterns/level05_logical_loop_combinations/Question033.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 05: Logical Loop Combinations` (File Question #33)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper compares the original value with its reversed form, because a palindrome reads the same both ways.
> - The `for` loop starts with `int number = 1`, keeps running while `number <= 500` is true, and updates using `number++`.
> - The condition `isPalindromeNumber(number)` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def reverseNumber(number):
    value = abs(number)
    reversed = 0
    while True:
        reversed = reversed * 10 + value % 10
        value //= 10
        if not (value > 0):
            break
    return -reversed if number < 0 else reversed

def isPalindromeNumber(number):
    return abs(number) == abs(reverseNumber(number))

def main():
    # Question 33: : Print all numbers that are palindromes between 1-500.
    for number in range(1, 500 + 1):
        if isPalindromeNumber(number):
            print(number)


if __name__ == "__main__":
    main()
```

### 🔹 [Q110 of 305] Find the smallest and largest digit in a given number.
- 📁 **File:** `phase02_looping_patterns/level05_logical_loop_combinations/Question035.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 05: Logical Loop Combinations` (File Question #35)

**💡 Logic & Explanation:**
> - The value stored in `smallest` is the input that the conditions or loops work on.
> - The loop uses `% 10` to take the last digit and `/= 10` to remove that digit, which is the standard digit-processing pattern.
> - The if/else checks test the important cases in order: `digit < smallest`, `digit > largest`.
> - Only the branch whose condition becomes true prints its message.


```python
def main():
    # Question 35: : Find the smallest and largest digit in a given number.
    number = 58231
    value = abs(number)
    smallest = 9
    largest = 0
    while True:
        digit = value % 10
        if digit < smallest:
            smallest = digit
        if digit > largest:
            largest = digit
        value //= 10
        if not (value > 0):
            break
    print("Smallest = " + str(smallest) + ", Largest = " + str(largest))


if __name__ == "__main__":
    main()
```

### 🔹 [Q111 of 305] Print all numbers from 1-n whose binary representation has an even number of 1s.
- 📁 **File:** `phase02_looping_patterns/level05_logical_loop_combinations/Question036.py`
- 🏷️ **Phase:** `Phase 02: Looping Patterns` | **Level/Category:** `Level 05: Logical Loop Combinations` (File Question #36)

**💡 Logic & Explanation:**
> - The value stored in `n` is the input that the conditions or loops work on.
> - The `for` loop starts with `int number = 1`, keeps running while `number <= n` is true, and updates using `number++`.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.
> - The if/else checks test the important cases in order: `value % 2 == 1`, `ones % 2 == 0`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 36: : Print all numbers from 1-n whose binary representation has an even number of 1s.
    n = 30
    for number in range(1, n + 1):
        value = number
        ones = 0
        while value > 0:
            if value % 2 == 1:
                ones += 1
            value /= 2
        if ones % 2 == 0:
            print(number)


if __name__ == "__main__":
    main()
```

- 🏷️ **Phase:** `Phase 03: Recursion` | **Level/Category:** `Level 01: Foundation Recursion` (File Question #1)

**💡 Logic & Explanation:**
> - The values `1, 5` passed to `printIncreasing` set the starting point and stopping limit for the recursive logic.
> - Inside `printIncreasing`, the condition `current > n` is the base case; it stops the repeated recursive calls.
> - The recursive call `printIncreasing(current + 1, n)` moves the work toward the base case by changing the argument each time.
> - One value is printed before the next recursive call, so the output is built step by step.


```python
def printIncreasing(current, n):
    if current > n:
        return
    print(current)
    printIncreasing(current + 1, n)

def main():
    # Question 1: : Print numbers from 1 to n using recursion.
    printIncreasing(1, 5)


if __name__ == "__main__":
    main()
```

> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper builds the answer from the last digit or last character toward the first one.
> - The calculated answer is printed after the logic produces the final value.


```python
def reverseNumber(number):
    value = abs(number)
    reversed = 0
    while True:
        reversed = reversed * 10 + value % 10
        value //= 10
        if not (value > 0):
            break
    return -reversed if number < 0 else reversed

def main():
    # Question 12: : Reverse a number recursively.
    number = 1234
    print("Reverse = " + str(reverseNumber(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q129 of 305] Find product of digits of a number recursively.
- 📁 **File:** `phase03_recursion/level02_number_based_recursion/Question014.py`
- 🏷️ **Phase:** `Phase 03: Recursion` | **Level/Category:** `Level 02: Number Based Recursion` (File Question #14)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The calculated answer is printed after the logic produces the final value.


```python
def productDigits(number):
    value = abs(number)
    product = 1
    if value == 0:
        return 0
    while value > 0:
        product *= value % 10
        value //= 10
    return product

def main():
    # Question 14: : Find product of digits of a number recursively.
    number = 2345
    print("Product of digits = " + str(productDigits(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q130 of 305] Find GCD (HCF) of two numbers using Euclid's algorithm recursively.
- 📁 **File:** `phase03_recursion/level02_number_based_recursion/Question015.py`
- 🏷️ **Phase:** `Phase 03: Recursion` | **Level/Category:** `Level 02: Number Based Recursion` (File Question #15)

**💡 Logic & Explanation:**

> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.
> - The calculated answer is printed after the logic produces the final value.


```python
def main():
    # Question 15: : Find GCD (HCF) of two numbers using Euclid's algorithm recursively.
    first = 48
    second = 18
    while second != 0:
        remainder = first % second
        first = second
        second = remainder
    print("GCD = " + str(abs(first)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q148 of 305] Count vowels in a string recursively.
- 📁 **File:** `phase03_recursion/level04_string_recursion/Question033.py`
- 🏷️ **Phase:** `Phase 03: Recursion` | **Level/Category:** `Level 04: String Recursion` (File Question #33)

**💡 Logic & Explanation:**
> - The value stored in `text` is the input that the conditions or loops work on.
> - The recursive return `(isVowel(text.charAt(index)) ? 1 : 0) + countVowelsRecursive(text, index + 1)` solves a smaller part first, then combines it with the current value.
> - The recursive call `countVowelsRecursive(text, index + 1)` moves the work toward the base case by changing the argument each time.
> - The calculated answer is printed after the logic produces the final value.


```python
def isVowel(ch):
    lower = ch.lower()
    return lower == 'a' or lower == 'e' or lower == 'i' or lower == 'o' or lower == 'u'

def countVowelsRecursive(text, index):
    if index == len(text):
        return 0
    return (1 if isVowel(text[index]) else 0) + countVowelsRecursive(text, index + 1)

def main():
    # Question 33: : Count vowels in a string recursively.
    text = "recursion"
    print("Vowels = " + str(countVowelsRecursive(text, 0)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q149 of 305] Remove all spaces from a string recursively.
- 📁 **File:** `phase03_recursion/level04_string_recursion/Question034.py`
- 🏷️ **Phase:** `Phase 03: Recursion` | **Level/Category:** `Level 04: String Recursion` (File Question #34)

**💡 Logic & Explanation:**
> - The value stored in `text` is the input that the conditions or loops work on.
> - The recursive return `(ch == ' ' ? "" : ch) + removeSpacesRecursive(text, index + 1)` solves a smaller part first, then combines it with the current value.
> - The recursive call `removeSpacesRecursive(text, index + 1)` moves the work toward the base case by changing the argument each time.
> - The calculated answer is printed after the logic produces the final value.


```python
def removeSpacesRecursive(text, index):
    if index == len(text):
        return ""
    ch = text[index]
    return ("" if ch == ' ' else ch) + removeSpacesRecursive(text, index + 1)

def main():
    # Question 34: : Remove all spaces from a string recursively.
    text = "lo g ic"
    print(removeSpacesRecursive(text, 0))


if __name__ == "__main__":
    main()
```

> - The recursive call `replaceCharRecursive(text, index + 1, oldChar, newChar)` moves the work toward the base case by changing the argument each time.
> - The calculated answer is printed after the logic produces the final value.


```python
def replaceCharRecursive(text, index, oldChar, newChar):
    if index == len(text):
        return ""
    ch = newChar if text[index] == oldChar else text[index]
    return ch + replaceCharRecursive(text, index + 1, oldChar, newChar)

def main():
    # Question 35: : Replace all occurrences of a character (say 'a' -> 'x') recursively.
    text = "banana"
    print(replaceCharRecursive(text, 0, 'a', 'x'))


if __name__ == "__main__":
    main()
```

### 🔹 [Q155 of 305] Count consonants and vowels separately using recursion.
- 📁 **File:** `phase03_recursion/level04_string_recursion/Question040.py`
- 🏷️ **Phase:** `Phase 03: Recursion` | **Level/Category:** `Level 04: String Recursion` (File Question #40)

**💡 Logic & Explanation:**
> - The value stored in `text` is the input that the conditions or loops work on.
> - The recursive return `(isVowel(text.charAt(index)) ? 1 : 0) + countVowelsRecursive(text, index + 1)` solves a smaller part first, then combines it with the current value.
> - The recursive call `countVowelsRecursive(text, index + 1)` moves the work toward the base case by changing the argument each time.
> - The calculated answer is printed after the logic produces the final value.


```python
def isVowel(ch):
    lower = ch.lower()
    return lower == 'a' or lower == 'e' or lower == 'i' or lower == 'o' or lower == 'u'

def countVowelsRecursive(text, index):
    if index == len(text):
        return 0
    return (1 if isVowel(text[index]) else 0) + countVowelsRecursive(text, index + 1)

def countConsonantsRecursive(text, index):
    if index == len(text):
        return 0
    ch = text[index]
    add = 1 if ch.isalpha() and not isVowel(ch) else 0
    return add + countConsonantsRecursive(text, index + 1)

def main():
    # Question 40: : Count consonants and vowels separately using recursion.
    text = "recursion"
    print("Vowels = " + str(countVowelsRecursive(text, 0)))
    print("Consonants = " + str(countConsonantsRecursive(text, 0)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q173 of 305] Find the count of prime numbers in the array.
- 📁 **File:** `phase04_basic_arrays/level02_searching_counting_logic/Question018.py`
- 🏷️ **Phase:** `Phase 04: Basic Arrays` | **Level/Category:** `Level 02: Searching Counting Logic` (File Question #18)

**💡 Logic & Explanation:**
> - The value stored in `numbers` is the input that the conditions or loops work on.
> - The helper tries possible divisors from 2 up to the square root of the number.
> - If any divisor divides the number exactly, the number is not prime; otherwise it is prime.
> - The enhanced `for` loop visits every element in the array one by one.
> - The variable `count` starts at 0 and increases only when the current item satisfies the question's condition.
> - The condition `isPrime(number)` decides whether the current value matches the requirement.


```python
def isPrime(number):
    if number <= 1:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True

def main():
    # Question 18: : Find the count of prime numbers in the array.
    numbers = [4, -2, 7, 0, 7, 3]
    count = 0
    for number in numbers:
        if isPrime(number):
            count += 1
    print("Prime count = " + str(count))


if __name__ == "__main__":
    main()
```

### 🔹 [Q181 of 305] Reverse an array (without using built-in reverse).
- 📁 **File:** `phase04_basic_arrays/level03_transformation_manipulation/Question026.py`
- 🏷️ **Phase:** `Phase 04: Basic Arrays` | **Level/Category:** `Level 03: Transformation Manipulation` (File Question #26)

**💡 Logic & Explanation:**
> - The variables `left`, `temp`, and `numbers` hold the values that the logic checks, counts, or transforms.
> - The helper builds the answer from the last digit or last character toward the first one.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.


```python
def printArray(values):
    for value in values:
        print(str(value) + " ", end="")
    print()

def main():
    # Question 26: : Reverse an array (without using built-in reverse).
    numbers = [4, -2, 7, 0, 7, 3]
    left = 0
    right = len(numbers) - 1
    while left < right:
        temp = numbers[left]
        numbers[left] = numbers[right]
        numbers[right] = temp
        left += 1
        right -= 1
printArray(numbers)


if __name__ == "__main__":
    main()
```

> - The helper calculates the required result and returns it to the print statement.

```python
def printArray(values):
    for value in values:
        print(str(value) + " ", end="")
    print()

def main():
    # Question 28: : Rotate an array by one position to the right.
    numbers = [4, -2, 7, 0, 7, 3]
    last = numbers[len(numbers) - 1]
    i = len(numbers) - 1
    while i > 0:
        numbers[i] = numbers[i - 1]
        i -= 1
    numbers[0] = last
    printArray(numbers)


if __name__ == "__main__":
    main()
```

### 🔹 [Q187 of 305] Compare two arrays - check if they contain the same elements (ignore order).
- 📁 **File:** `phase04_basic_arrays/level04_aggregate_comparative/Question032.py`
- 🏷️ **Phase:** `Phase 04: Basic Arrays` | **Level/Category:** `Level 04: Aggregate Comparative` (File Question #32)

**💡 Logic & Explanation:**
> - The variables `same`, `first`, `second`, and `used` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < first.length && same` is true, and updates using `i++`.
> - The boolean `found` starts as false and becomes true as soon as the searched value is seen.
> - The if/else checks test the important cases in order: `!used[j] && first[i] == second[j]`, `!found`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 32: : Compare two arrays - check if they contain the same elements (ignore order).
    first = [1, 2, 3]
    second = [3, 2, 1]
    same = len(first) == len(second)
    used = [False] * (len(second))
    for i in range(0, len(first) and same):
        found = False
        for j in range(0, len(second)):
            if not used[j] and first[i] == second[j]:
                used[j] = True
                found = True
                break
        if not found:
            same = False
    print("Same elements" if same else "Different elements")


if __name__ == "__main__":
    main()
```

### 🔹 [Q194 of 305] Create a frequency array of numbers (count occurrence of each number).

> - The variables `numbers`, and `visited` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < numbers.length` is true, and updates using `i++`.
> - The condition `numbers[i] == numbers[j]` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 39: : Create a frequency array of numbers (count occurrence of each number).
    numbers = [4, -2, 7, 0, 7, 3]
    visited = [False] * (len(numbers))
    for i in range(0, len(numbers)):
        if visited[i]:
            continue
        count = 1
        for j in range(i + 1, len(numbers)):
            if numbers[i] == numbers[j]:
                visited[j] = True
                count += 1
print(str(numbers[i]) + " -> " + str(count))


if __name__ == "__main__":
    main()
```

### 🔹 [Q204 of 305] Print the frequency of each distinct element.
- 📁 **File:** `phase04_basic_arrays/level05_logical_applied_arrays/Question049.py`
- 🏷️ **Phase:** `Phase 04: Basic Arrays` | **Level/Category:** `Level 05: Logical Applied Arrays` (File Question #49)

**💡 Logic & Explanation:**
> - The variables `numbers`, and `visited` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < numbers.length` is true, and updates using `i++`.
> - The if/else checks test the important cases in order: `visited[i]`, `numbers[i] == numbers[j]`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 49: : Print the frequency of each distinct element.
    numbers = [4, 7, 4, 2, 7, 9, 2]
    visited = [False] * (len(numbers))

    for i in range(0, len(numbers)):
        if visited[i]:
            continue

        count = 1
        for j in range(i + 1, len(numbers)):
            if numbers[i] == numbers[j]:
                count += 1
                visited[j] = True

        print(str(numbers[i]) + " appears " + str(count) + " times")


if __name__ == "__main__":
    main()
```

### 🔹 [Q205 of 305] Print all unique elements (those that occur exactly once).
- 📁 **File:** `phase04_basic_arrays/level05_logical_applied_arrays/Question050.py`
- 🏷️ **Phase:** `Phase 04: Basic Arrays` | **Level/Category:** `Level 05: Logical Applied Arrays` (File Question #50)

**💡 Logic & Explanation:**
> - The value stored in `numbers` is the input that the conditions or loops work on.
> - The enhanced `for` loop visits every element in the array one by one.
> - The variable `count` starts at 0 and increases only when the current item satisfies the question's condition.
> - The if/else checks test the important cases in order: `value == numbers[i]`, `count == 1`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 50: : Print all unique elements (those that occur exactly once).
    numbers = [4, -2, 7, 0, 7, 3]
    for i in range(0, len(numbers)):
        count = 0
        for value in numbers:
            if value == numbers[i]:
                count += 1
        if count == 1:
            print(numbers[i])


if __name__ == "__main__":
    main()
```

### 🔹 [Q216 of 305] Count how many vowels and consonants are in a string.
- 📁 **File:** `phase05_strings/category02_counting_character_analysis/Question011.py`
- 🏷️ **Phase:** `Phase 05: Strings` | **Level/Category:** `Category 02: Counting Character Analysis` (File Question #11)

**💡 Logic & Explanation:**
> - The variables `text`, `vowels`, and `ch` hold the values that the logic checks, counts, or transforms.
> - The helper checks characters against `a`, `e`, `i`, `o`, and `u`, so the main loop only needs to use that true/false result.
> - The `for` loop starts with `int i = 0`, keeps running while `i < text.length()` is true, and updates using `i++`.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The if/else checks test the important cases in order: `Character.isLetter(ch)`, `isVowel(ch)`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def isVowel(ch):
    lower = ch.lower()
    return lower == 'a' or lower == 'e' or lower == 'i' or lower == 'o' or lower == 'u'

def main():
    # Question 11: : Count how many vowels and consonants are in a string.
    text = "logic"
    vowels = 0
    consonants = 0
    for i in range(0, len(text)):
        ch = text[i]
        if ch.isalpha():
            if isVowel(ch):
                vowels += 1
            else:
                consonants += 1
    print("Vowels = " + str(vowels) + ", Consonants = " + str(consonants))


if __name__ == "__main__":
    main()
```

### 🔹 [Q219 of 305] Find the frequency of each character in a string (without using a map).
- 📁 **File:** `phase05_strings/category02_counting_character_analysis/Question014.py`
- 🏷️ **Phase:** `Phase 05: Strings` | **Level/Category:** `Category 02: Counting Character Analysis` (File Question #14)

**💡 Logic & Explanation:**
> - The variables `text`, and `visited` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < text.length()` is true, and updates using `i++`.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The condition `text.charAt(i) == text.charAt(j)` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 14: : Find the frequency of each character in a string (without using a map).
    text = "banana"
    visited = [False] * (len(text))
    for i in range(0, len(text)):
        if visited[i]:
            continue
        count = 1
        for j in range(i + 1, len(text)):
            if text[i] == text[j]:
                visited[j] = True
                count += 1
        print(str(text[i]) + " -> " + str(count))


if __name__ == "__main__":
    main()
```

### 🔹 [Q222 of 305] Count how many alphabets are before 'm' and after 'm' in a given string.
- 📁 **File:** `phase05_strings/category02_counting_character_analysis/Question017.py`
- 🏷️ **Phase:** `Phase 05: Strings` | **Level/Category:** `Category 02: Counting Character Analysis` (File Question #17)

**💡 Logic & Explanation:**
> - The variables `text`, `before`, and `ch` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < text.length()` is true, and updates using `i++`.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The if/else checks test the important cases in order: `ch >= 'a' && ch < 'm'`, `ch > 'm' && ch <= 'z'`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 17: : Count how many alphabets are before 'm' and after 'm' in a given string.
    text = "alphabet"
    before = 0
    after = 0
    for i in range(0, len(text)):
        ch = text[i].lower()
        if ch >= 'a' and ch < 'm':
            before += 1
        elif ch > 'm' and ch <= 'z':
            after += 1
    print("Before m = " + str(before) + ", After m = " + str(after))


if __name__ == "__main__":
    main()
```

> - The variables `text`, `start`, and `end` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int start = 0`, keeps running while `start < text.length()` is true, and updates using `start++`.
> - The variable `count` starts at 0 and increases only when the current item satisfies the question's condition.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The condition `text.charAt(start) == text.charAt(end)` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 18: : Count how many substrings start and end with the same character (simple logic).
    text = "abca"
    count = 0
    for start in range(0, len(text)):
        for end in range(start, len(text)):
            if text[start] == text[end]:
                count += 1
    print("Count = " + str(count))


if __name__ == "__main__":
    main()
```

### 🔹 [Q226 of 305] Reverse a string without using built-in reverse.

- 🏷️ **Phase:** `Phase 05: Strings` | **Level/Category:** `Category 03: Reversing Palindromic` (File Question #21)

**💡 Logic & Explanation:**
> - The value stored in `text` is the input that the conditions or loops work on.
> - The helper builds the answer from the last digit or last character toward the first one.
> - The calculated answer is printed after the logic produces the final value.


```python
def reverseString(text):
    builder = ""
    for i in range(len(text) - 1, 0 - 1, -1):
        builder += str(text[i])
    return builder.toString()

def main():
    # Question 21: : Reverse a string without using built-in reverse.
    text = "logic"
    print(reverseString(text))


if __name__ == "__main__":
    main()
```

### 🔹 [Q229 of 305] Check whether a string is a palindrome.
- 📁 **File:** `phase05_strings/category03_reversing_palindromic/Question024.py`
- 🏷️ **Phase:** `Phase 05: Strings` | **Level/Category:** `Category 03: Reversing Palindromic` (File Question #24)

**💡 Logic & Explanation:**
> - The value stored in `text` is the input that the conditions or loops work on.
> - The helper compares the original value with its reversed form, because a palindrome reads the same both ways.
> - The calculated answer is printed after the logic produces the final value.


```python
def isPalindromeString(text):
    cleaned = text.replace(" ", "").lower()
    left = 0
    right = len(cleaned) - 1
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    return True

def main():
    # Question 24: : Check whether a string is a palindrome.
    text = "level"
    print("Palindrome" if isPalindromeString(text) else "Not palindrome")


if __name__ == "__main__":
    main()
```

### 🔹 [Q244 of 305] Swap case: uppercase -> lowercase and lowercase -> uppercase.

**💡 Logic & Explanation:**
> - The variables `text`, and `ch` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < text.length()` is true, and updates using `i++`.
> - A `StringBuilder` is used so characters or words can be added gradually while building the final string.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The if/else checks test the important cases in order: `Character.isUpperCase(ch)`, `Character.isLowerCase(ch)`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 39: : Swap case: uppercase -> lowercase and lowercase -> uppercase.
    text = "Java Logic"
    result = ""

    for i in range(0, len(text)):
        ch = text[i]
        if ch.isupper():
            result += str(ch.lower())
        elif ch.islower():
            result += str(ch.upper())
        else:
            result += str(ch)

    print(result)


if __name__ == "__main__":
    main()
```

### 🔹 [Q257 of 305] Find the sum of digits of a number (use loop).
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question002.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #2)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The calculated answer is printed after the logic produces the final value.


```python
def sumDigits(number):
    value = abs(number)
    sum = 0
    while True:
        sum += value % 10
        value //= 10
        if not (value > 0):
            break
    return sum

def main():
    # Question 2: : Find the sum of digits of a number (use loop).
    number = 9876
    print("Sum of digits = " + str(sumDigits(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q258 of 305] Check if a number is an Armstrong number.
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question003.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #3)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper counts how many digits the number has by repeatedly dividing a copy of the number by 10.
> - Then it extracts each digit with `% 10`, raises that digit to the digit-count power, and adds it to `sum`.
> - If the final `sum` equals the original number, the helper returns true.
> - The calculated answer is printed after the logic produces the final value.


```python
def isArmstrong(number):
    value = number
    digits = 0
    while True:
        digits += 1
        value //= 10
        if not (value > 0):
            break
    sum = 0
    value = number
    while True:
        digit = value % 10
        power = 1
        for i in range(1, digits + 1):
            power *= digit
        sum += power
        value //= 10
        if not (value > 0):
            break
    return sum == number

def main():
    # Question 3: : Check if a number is an Armstrong number.
    number = 153
    print("Armstrong number" if isArmstrong(number) else "Not Armstrong number")


if __name__ == "__main__":
    main()
```

### 🔹 [Q259 of 305] Print all Armstrong numbers between 1 and 1000.
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question004.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #4)

**💡 Logic & Explanation:**
> - The variables `number`, `value`, `sum`, and `digit` hold the values that the logic checks, counts, or transforms.
> - The helper counts how many digits the number has by repeatedly dividing a copy of the number by 10.
> - Then it extracts each digit with `% 10`, raises that digit to the digit-count power, and adds it to `sum`.
> - If the final `sum` equals the original number, the helper returns true.
> - The loop uses `% 10` to take the last digit and `/= 10` to remove that digit, which is the standard digit-processing pattern.
> - The variable `sum` starts at 0 and collects values as the loop finds numbers that should be included.


```python
def factorial(number):
    result = 1
    for value in range(2, number + 1):
        result *= value
    return result

def main():
    # Question 4: : Print all Armstrong numbers between 1 and 1000.
    number = 145
    value = number
    sum = 0
    while value > 0:
        digit = value % 10
        sum += factorial(digit)
        value //= 10
    print("Strong number" if sum == number else "Not strong number")


if __name__ == "__main__":
    main()
```

### 🔹 [Q260 of 305] Find the factorial of a number using recursion.
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question005.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #5)

**💡 Logic & Explanation:**
> - The value stored in `number` is the number whose factorial has to be calculated.
> - The base case returns 1 when the number reaches 0 or 1, because factorial stops there.
> - The recursive step `number * factorial(number - 1)` multiplies the current number by the factorial of the smaller number.
> - When all recursive calls return, the final multiplied value is printed as the factorial.


```python
def factorial(number):
    if number == 0 or number == 1:
        return 1
    return number * factorial(number - 1)

def main():
    # Question 5: : Find the factorial of a number using recursion.
    number = 5
    print("Factorial = " + str(factorial(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q262 of 305] Print all prime numbers between 1 and N.
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question007.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #7)

**💡 Logic & Explanation:**
> - The value stored in `n` is the input that the conditions or loops work on.
> - The helper tries possible divisors from 2 up to the square root of the number.
> - If any divisor divides the number exactly, the number is not prime; otherwise it is prime.
> - The `for` loop starts with `int number = 2`, keeps running while `number <= n` is true, and updates using `number++`.
> - The condition `isPrime(number)` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def isPrime(number):
    if number <= 1:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True

def main():
    # Question 7: : Print all prime numbers between 1 and N.
    n = 100
    for number in range(2, n + 1):
        if isPrime(number):
            print(number)


if __name__ == "__main__":
    main()
```

### 🔹 [Q263 of 305] Print the reverse of a number (123 -> 321).
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question008.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #8)

**💡 Logic & Explanation:**
> - The value stored in `number` is the input that the conditions or loops work on.
> - The helper builds the answer from the last digit or last character toward the first one.
> - The calculated answer is printed after the logic produces the final value.


```python
def reverseNumber(number):
    value = abs(number)
    reversed = 0
    while True:
        reversed = reversed * 10 + value % 10
        value //= 10
        if not (value > 0):
            break
    return -reversed if number < 0 else reversed

def main():
    # Question 8: : Print the reverse of a number (123 -> 321).
    number = 12345
    print("Reverse = " + str(reverseNumber(number)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q264 of 305] Check if a number is palindrome (121 -> true).
- 📁 **File:** `phase06_mixed_logical_challenges/category01_number_based_combinations/Question009.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 01: Number Based Combinations` (File Question #9)

**💡 Logic & Explanation:**

> - The loop uses `% 10` to take the last digit and `/= 10` to remove that digit, which is the standard digit-processing pattern.
> - The calculated answer is printed after the logic produces the final value.


```python
def main():
    # Question 9: : Check if a number is palindrome (121 -> true).
    number = 121
    value = number
    reversed = 0

    while value > 0:
        digit = value % 10
        reversed = reversed * 10 + digit
        value //= 10

    print("Palindrome" if number == reversed else "Not palindrome")


if __name__ == "__main__":
    main()
```

### 🔹 [Q266 of 305] Check if two strings are anagrams (without using collections).
- 📁 **File:** `phase06_mixed_logical_challenges/category02_string_logic_mix/Question011.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 02: String Logic Mix` (File Question #11)

**💡 Logic & Explanation:**
> - The variables `first`, `second`, and `anagram` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < first.length()` is true, and updates using `i++`.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The condition `value != 0` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 11: : Check if two strings are anagrams (without using collections).
    first = "listen"
    second = "silent"
    frequency = [0] * (256)
    for i in range(0, len(first)):
        frequency[first.charAt(i)] += 1
    for i in range(0, len(second)):
        frequency[second.charAt(i)] -= 1
    anagram = len(first) == len(second)
    for value in frequency:
        if value != 0:
            anagram = False
    print("Anagram" if anagram else "Not anagram")


if __name__ == "__main__":
    main()
```

### 🔹 [Q267 of 305] Count vowels in each word of a sentence.
- 📁 **File:** `phase06_mixed_logical_challenges/category02_string_logic_mix/Question012.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 02: String Logic Mix` (File Question #12)

**💡 Logic & Explanation:**
> - The value stored in `sentence` is the input that the conditions or loops work on.
> - The helper checks characters against `a`, `e`, `i`, `o`, and `u`, so the main loop only needs to use that true/false result.
> - The loop goes word by word after splitting the sentence, so each word can be checked separately.
> - `split("\\s+")` breaks the sentence wherever spaces appear, giving us separate words to process.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def isVowel(ch):
    lower = ch.lower()
    return lower == 'a' or lower == 'e' or lower == 'i' or lower == 'o' or lower == 'u'

def countVowels(text):
    count = 0
    for i in range(0, len(text)):
        if isVowel(text[i]):
            count += 1
    return count

def main():
    # Question 12: : Count vowels in each word of a sentence.
    sentence = "logic building"
    for word in sentence.split():
        print(str(word) + " -> " + str(countVowels(word)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q270 of 305] Print characters that appear more than once (without map).

**💡 Logic & Explanation:**

> - `charAt(i)` lets the code inspect one character at a time by index.

### 🔹 [Q275 of 305] Remove duplicate words from a sentence.
- 📁 **File:** `phase06_mixed_logical_challenges/category02_string_logic_mix/Question020.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 02: String Logic Mix` (File Question #20)

**💡 Logic & Explanation:**
> - The variables `seen`, and `words` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < words.length` is true, and updates using `i++`.
> - A `StringBuilder` is used so characters or words can be added gradually while building the final string.
> - `split("\\s+")` breaks the sentence wherever spaces appear, giving us separate words to process.
> - The condition `words[i].equals(words[j])` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 20: : Remove duplicate words from a sentence.
    words = "java logic java code".split()
    result = ""
    for i in range(0, len(words)):
        seen = False
        for j in range(0, i):
            if words[i].equals(words[j]):
                seen = True
        if not seen:
            result += str(words[i]).append(" ")
    print(result.toString().strip())


if __name__ == "__main__":
    main()
```

### 🔹 [Q276 of 305] Find the maximum and minimum element in an array.
- 📁 **File:** `phase06_mixed_logical_challenges/category03_array_looping_logic/Question021.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 03: Array Looping Logic` (File Question #21)

**💡 Logic & Explanation:**
> - The value stored in `numbers` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The calculated answer is printed after the logic produces the final value.


```python
def max(values):
    answer = values[0]
    for value in values:
        if value > answer:
            answer = value
    return answer

def min(values):
    answer = values[0]
    for value in values:
        if value < answer:
            answer = value
    return answer

def main():
    # Question 21: : Find the maximum and minimum element in an array.
    numbers = [4, -2, 7, 0, 7, 3]
    print("Maximum = " + str(max(numbers)))
    print("Minimum = " + str(min(numbers)))


if __name__ == "__main__":
    main()
```

### 🔹 [Q279 of 305] Reverse an array in-place.
- 📁 **File:** `phase06_mixed_logical_challenges/category03_array_looping_logic/Question024.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 03: Array Looping Logic` (File Question #24)

**💡 Logic & Explanation:**
> - The variables `left`, `temp`, and `numbers` hold the values that the logic checks, counts, or transforms.
> - The helper builds the answer from the last digit or last character toward the first one.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.


```python
def printArray(values):
    for value in values:
        print(str(value) + " ", end="")
    print()

def main():
    # Question 24: : Reverse an array in-place.
    numbers = [4, -2, 7, 0, 7, 3]
    left = 0
    right = len(numbers) - 1
    while left < right:
        temp = numbers[left]
        numbers[left] = numbers[right]
        numbers[right] = temp
        left += 1
        right -= 1
printArray(numbers)


if __name__ == "__main__":
    main()
```

### 🔹 [Q280 of 305] Shift all zeros to the end of the array.
- 📁 **File:** `phase06_mixed_logical_challenges/category03_array_looping_logic/Question025.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 03: Array Looping Logic` (File Question #25)

**💡 Logic & Explanation:**
> - The variables `index`, and `numbers` hold the values that the logic checks, counts, or transforms.
> - The helper calculates the required result and returns it to the print statement.
> - The enhanced `for` loop visits every element in the array one by one.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.


```python
def printArray(values):
    for value in values:
        print(str(value) + " ", end="")
    print()

def main():
    # Question 25: : Shift all zeros to the end of the array.
    numbers = [0, 4, 0, 5, 7, 0]
    index = 0
    for number in numbers:
        if number != 0:
            numbers[index] = number
            index += 1
    while index < len(numbers):
        numbers[index] = 0
        index += 1
    printArray(numbers)


if __name__ == "__main__":
    main()
```

> - The helper calculates the required result and returns it to the print statement.
> - The `for` loop starts with `int i = 0`, keeps running while `i < first.length` is true, and updates using `i++`.


```python
def printArray(values):
    for value in values:
        print(str(value) + " ", end="")
    print()

def main():
    # Question 27: : Merge two arrays into one.
    first = [1, 2, 3]
    second = [4, 5]
    merged = [0] * (len(first) + len(second))
    for i in range(0, len(first)):
        merged[i] = first[i]
    for i in range(0, len(second)):
        merged[len(first) + i] = second[i]
    printArray(merged)


if __name__ == "__main__":
    main()
```

```python
def main():
    # Question 35: : Count how many times a number appears consecutively in an array.
    numbers = [2, 2, 2, 3, 3, 4]
    target = 2
    maxRun = 0
    currentRun = 0
    for number in numbers:
        if number == target:
            currentRun += 1
        else:
            currentRun = 0
        if currentRun > maxRun:
            maxRun = currentRun
    print("Longest consecutive count = " + str(maxRun))


if __name__ == "__main__":
    main()
```

> - `System.out.println()` is used only to move to the next line after the pattern output is complete.


```python
def main():
    # Question 38: : Print Pascal's triangle up to N rows.
    rows = 5
    for row in range(0, rows):
        value = 1
        for col in range(0, row + 1):
            print(str(value) + " ", end="")
            value = value * (row - col) / (col + 1)
        print()


if __name__ == "__main__":
    main()
```

### 🔹 [Q294 of 305] Generate Fibonacci series up to N using recursion.
- 📁 **File:** `phase06_mixed_logical_challenges/category04_nested_logic_pattern_flow/Question039.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 04: Nested Logic Pattern Flow` (File Question #39)

**💡 Logic & Explanation:**
> - The value stored in `terms` is the input that the conditions or loops work on.
> - Inside `fibonacci`, the condition `n == 0` is the base case; it stops the repeated recursive calls.
> - The recursive return `fibonacci(n - 1) + fibonacci(n - 2)` solves a smaller part first, then combines it with the current value.
> - The recursive call `fibonacci(n - 2)` moves the work toward the base case by changing the argument each time.
> - The `for` loop starts with `int i = 0`, keeps running while `i < terms` is true, and updates using `i++`.
> - `System.out.println()` is used only to move to the next line after the pattern output is complete.


```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

def main():
    # Question 39: : Generate Fibonacci series up to N using recursion.
    terms = 8

    for i in range(0, terms):
        print(str(fibonacci(i)) + " ", end="")
    print()


if __name__ == "__main__":
    main()
```

### 🔹 [Q295 of 305] Question 40
- 📁 **File:** `phase06_mixed_logical_challenges/category04_nested_logic_pattern_flow/Question040.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 04: Nested Logic Pattern Flow` (File Question #40)

**💡 Logic & Explanation:**
> - The value stored in `n` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The outer loop controls each row of the pattern, and the inner loop controls what gets printed inside that row.
> - The `while` or `do-while` loop keeps repeating until the changing value reaches the stopping condition.


```python
def main():
    # Question: Print spiral number matrix up to N x N.
    n = 4
    matrix = [[0 for _ in range(n)] for _ in range(n)]
    value = 1
    top, bottom, left, right = 0, n - 1, 0, n - 1
    while top <= bottom and left <= right:
        for col in range(left, right + 1):
            matrix[top][col] = value; value += 1
        top += 1
        for row in range(top, bottom + 1):
            matrix[row][right] = value; value += 1
        right -= 1
        for col in range(right, left - 1, -1):
            matrix[bottom][col] = value; value += 1
        bottom -= 1
        for row in range(bottom, top - 1, -1):
            matrix[row][left] = value; value += 1
        left += 1
    for row in matrix:
        print(*row)


if __name__ == "__main__":
    main()
```

### 🔹 [Q297 of 305] Take age inputs and count how many are adults, minors, seniors.
- 📁 **File:** `phase06_mixed_logical_challenges/category05_applied_reasoning/Question042.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 05: Applied Reasoning` (File Question #42)

**💡 Logic & Explanation:**
> - The variables `minors`, `adults`, `seniors`, and `ages` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int age : ages) { if (age < 18) { minors++`, keeps running while `} else if (age >= 60) { seniors++` is true, and updates using `} else { adults++; } } System.out.println("Minors = " + minors`.
> - The if/else checks test the important cases in order: `age < 18`, `age >= 60`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 42: : Take age inputs and count how many are adults, minors, seniors.
    ages = [12, 18, 35, 61, 8, 75]
    minors = 0
    adults = 0
    seniors = 0

    for age in ages:
        if age < 18:
            minors += 1
        elif age >= 60:
            seniors += 1
        else:
            adults += 1

    print("Minors = " + str(minors))
    print("Adults = " + str(adults))
    print("Seniors = " + str(seniors))


if __name__ == "__main__":
    main()
```

### 🔹 [Q298 of 305] Validate a password (at least one uppercase, lowercase, digit, special char).
- 📁 **File:** `phase06_mixed_logical_challenges/category05_applied_reasoning/Question043.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 05: Applied Reasoning` (File Question #43)

**💡 Logic & Explanation:**
> - The variables `password`, `hasDigit`, and `ch` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `int i = 0`, keeps running while `i < password.length()` is true, and updates using `i++`.
> - `charAt(i)` lets the code inspect one character at a time by index.
> - The if/else checks test the important cases in order: `Character.isDigit(ch)`, `Character.isUpperCase(ch)`, `Character.isLowerCase(ch)`.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 43: : Validate a password (at least one uppercase, lowercase, digit, special char).
    password = "Code123not "
    hasDigit = False
    hasUpper = False
    hasLower = False
    hasSpecial = False
    for i in range(0, len(password)):
        ch = password[i]
        if ch.isdigit():
            hasDigit = True
        elif ch.isupper():
            hasUpper = True
        elif ch.islower():
            hasLower = True
        else:
            hasSpecial = True
    if len(password) >= 8 and hasDigit and hasUpper and hasLower and hasSpecial:
        print("Valid password")
    else:
        print("Invalid password")


if __name__ == "__main__":
    main()
```

### 🔹 [Q300 of 305] Count how many times a coin lands on heads/tails (use random).
- 📁 **File:** `phase06_mixed_logical_challenges/category05_applied_reasoning/Question045.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 05: Applied Reasoning` (File Question #45)

**💡 Logic & Explanation:**
> - The variables `heads`, and `results` hold the values that the logic checks, counts, or transforms.
> - The `for` loop starts with `String result : results) { if (result.equals("H")) heads++`, keeps running while `else tails++` is true, and updates using `} System.out.println("Heads = " + heads + ", Tails = " + tails`.
> - The condition `result.equals("H")` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 45: : Count how many times a coin lands on heads/tails (use random).
    results = {"H", "T", "H", "H", "T"}
    heads = 0
    tails = 0
    for result in results:
        if result.equals("H"):
            heads += 1
        else:
            tails += 1
    print("Heads = " + str(heads) + ", Tails = " + str(tails))


if __name__ == "__main__":
    main()
```

### 🔹 [Q301 of 305] Print frequency of each digit in a number.
- 📁 **File:** `phase06_mixed_logical_challenges/category05_applied_reasoning/Question046.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 05: Applied Reasoning` (File Question #46)

**💡 Logic & Explanation:**
> - We store the sample number in `number` and create a `frequency` array with 10 positions, one for each digit from 0 to 9.
> - The `while` loop extracts the last digit using `% 10` and then removes it using `/= 10`.
> - For every extracted digit, `frequency[digit]++` increases the counter for that exact digit.
> - The final `for` loop goes through the frequency array and prints only the digits that appeared at least once.


```python
def main():
    # Question 46: : Print frequency of each digit in a number.
    number = 122333045
    frequency = [0] * (10)
    value = number

    while value > 0:
        digit = value % 10
        frequency[digit] += 1
        value //= 10

    for digit in range(0, len(frequency)):
        if frequency[digit] > 0:
            print(str(digit) + " appears " + str(frequency[digit]) + " times")


if __name__ == "__main__":
    main()
```

> - The loop repeats the same check or calculation for every required value.
> - The condition `a == b` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 47: : Find common elements between two arrays.
    first = [1, 2, 3, 4]
    second = [3, 4, 5]
    for a in first:
        for b in second:
            if a == b:
                print(a)


if __name__ == "__main__":
    main()
```

> - The condition `second.indexOf(ch) >= 0 && first.indexOf(ch) == i` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def main():
    # Question 48: : Print characters that are common in two strings.
    first = "logic"
    second = "coding"
    for i in range(0, len(first)):
        ch = first[i]
        if second.find(ch) >= 0 and first.find(ch) == i:
            print(ch)


if __name__ == "__main__":
    main()
```

### 🔹 [Q305 of 305] Print all palindromic words from a sentence.
- 📁 **File:** `phase06_mixed_logical_challenges/category05_applied_reasoning/Question050.py`
- 🏷️ **Phase:** `Phase 06: Mixed Logical Challenges` | **Level/Category:** `Category 05: Applied Reasoning` (File Question #50)

**💡 Logic & Explanation:**
> - The value stored in `sentence` is the input that the conditions or loops work on.
> - The helper calculates the required result and returns it to the print statement.
> - The loop goes word by word after splitting the sentence, so each word can be checked separately.
> - `split("\\s+")` breaks the sentence wherever spaces appear, giving us separate words to process.
> - The condition `isPalindromeString(word)` decides whether the current value matches the requirement.
> - The print statement shows each value as soon as that value is calculated inside the loop.


```python
def isPalindromeString(text):
    cleaned = text.replace(" ", "").lower()
    left = 0
    right = len(cleaned) - 1
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    return True

def main():
    # Question 50: : Print all palindromic words from a sentence.
    sentence = "madam and level are words"
    for word in sentence.split():
        if isPalindromeString(word):
            print(word)


if __name__ == "__main__":
    main()
```
