```python
# wap to calculate the square of numbers using multithreading 

import threading
numbers = [2, 3, 4, 5]

def calculate_square(num):
    print(f'The square of {num} is {num * num}')
            
t1 = threading.Thread(target=calculate_square, args=(numbers[0],))
t2 = threading.Thread(target=calculate_square, args=(numbers[1],))
t3 = threading.Thread(target=calculate_square, args=(numbers[2],))
t4 = threading.Thread(target=calculate_square, args=(numbers[3],))

t1.start()
t2.start()
t3.start()
t4.start()

t1.join()
t2.join()
t3.join()
t4.join()
```

```python
# wap to reverse multiple strings using multithreading

import threading
strings = ['python','java','sql']
def reverse_string(s):
    print(f'The reverse of {s} is {s[::-1]}')

t1 = threading.Thread(target=reverse_string, args=(strings[0],))
t2 = threading.Thread(target=reverse_string, args=(strings[1],))
t3 = threading.Thread(target=reverse_string, args=(strings[2],))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()
```

```python
# wap to calculate factorial of numbers using multithreading
import threading
numbers = [3, 4, 5]
def calculate_factorial(num):
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    print(f'The factorial of {num} is {factorial}')

t1 = threading.Thread(target=calculate_factorial, args=(numbers[0],))
t2 = threading.Thread(target=calculate_factorial, args=(numbers[1],))
t3 = threading.Thread(target=calculate_factorial, args=(numbers[2],))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()
```

```python
# wap to perform check balance, deposit, withdraw operations using single level inheritance with exceptions handling validation

class Bank:
    def __init__(self, accno, balance):
        self.accno = accno
        self.balance = balance
    
    def account_details(self):
        print(f'Account Number: {self.accno}')
        print(f'Balance: {self.balance}')

class SavingsAccount(Bank):
    def deposit(self, amount):
        try:
            if amount <= 0:
                raise ValueError("Deposit amount must be positive.")
            self.balance += amount
            print(f'Deposited: {amount}. New Balance: {self.balance}')
        except ValueError as e:
            print(f'deposit error: {e}')

    def withdraw(self, amount):
        try:
            if amount <= 0:
                raise ValueError("Withdrawal amount must be positive.")
            if amount > self.balance:
                raise ValueError("Insufficient balance.")
            self.balance -= amount
            print(f'Withdrawn: {amount}. New Balance: {self.balance}')
        except ValueError as e:
            print(f'withdrawal error: {e}')

s = SavingsAccount("123456789", 1000)
s.account_details()
s.deposit(500)
s.withdraw(200)
s.withdraw(1500)  # This will trigger insufficient balance error
```

```python
# wap to display bankname ,balanace , and perform deposit and withrwal 

class Bank:
    def bankname (self):
        self.bname = 'BOI'
        print(f'Bank Name: {self.bname}')

class Account(Bank):
    def __init__(self, accno, balance):
        self.accno = accno
        self.balance = balance
    
    def account_details(self):
        print(f'Account Number: {self.accno}')
        print(f'Balance: {self.balance}')

class transaction(Account):
    def deposit(self,amount):
        if amount > 0:
            self.balance += amount
            print(f'Deposited: {amount}. New Balance: {self.balance}')
        else:
            print("Invalid deposit amount. Please enter a positive value.")

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                print(f'Withdrawn: {amount}. New Balance: {self.balance}')
            else:
                print("Insufficient balance.")
        else:
            print("Invalid withdrawal amount. Please enter a positive value.")

t = transaction("123456789", 1000)
t.bankname()
t.account_details()
t.deposit(500)
t.withdraw(200)
t.withdraw(1500)  # This will trigger insufficient balance error
```

```python
# wap to display the employee name , empno, salary, department name and project name using the muliple level inheritance

class Employee:
    def __init__(self, empno, name, salary):
        self.empno = empno
        self.name = name
        self.salary = salary
    
    def display_employee_details(self):
        print(f'Employee Number: {self.empno}')
        print(f'Employee Name: {self.name}')
        print(f'Salary: {self.salary}')

class Department(Employee):
    def __init__(self, empno, name, salary, dept_name):
        super().__init__(empno, name, salary)
        self.dept_name = dept_name
    
    def display_department_details(self):
        print(f'Department Name: {self.dept_name}')

class Project(Department):
    def __init__(self, empno, name, salary, dept_name, pid, project_name):
        super().__init__(empno, name, salary, dept_name)
        self.pid = pid
        self.project_name = project_name
    
    def display_project_details(self):
        self.display_employee_details()
        self.display_department_details()
        print(f'Project ID: {self.pid}')
        print(f'Project Name: {self.project_name}')

p = Project(101, "Alice", 50000, "IT", "P001", "Project X")
p.display_project_details()
```

```python
# example using classname, methodname to achieve parent class properties

class A:
    def method_a(self):
        print("This is method A from class A")
class B(A):
    def method_b(self):
        print("This is method B from class B")
class C(B):
    def method_c(self):
        print("This is method C from class C")
        B.method_b(self)
        A.method_a(self)
obj = C()
obj.method_c()
print(C.__mro__)
```

```python

# Bank scenario example for multiple inheritance

class CheckBalance: # parent class 1
    def show_balance(self):
        print(f"Current Balance: {self.balance}")

class Deposit: # parent class 2
    def deposit_amount(self, amount):
        if amount <= 0:
            print("Invalid deposit amount! Amount must be greater than 0.")
        else:
            self.balance += amount
            print(f"Deposited Amount: {amount}")
            print(f"Updated Balance after Deposit: {self.balance}")


class Withdraw: # parent class 3
    def withdraw_amount(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount! Amount must be greater than 0.")
        elif amount > self.balance:
            print("Insufficient balance! Withdrawal failed.")
        else:
            self.balance -= amount
            print(f"Withdrawn Amount: {amount}")
            print(f"Updated Balance after Withdrawal: {self.balance}")


class Bank(CheckBalance, Deposit, Withdraw): # child class inheriting from all three parent classes
    def __init__(self, cname, acctno, balance):
        self.cname = cname
        self.acctno = acctno
        self.balance = balance

    def display_details(self):
        print(f"Customer Name : {self.cname}")
        print(f"Account Number: {self.acctno}")
        print(f"Opening Balance: {self.balance}")



customer = Bank('smith', 1234, 98000)

customer.display_details()
customer.show_balance()
customer.deposit_amount(2000)
customer.withdraw_amount(3000)


print(Bank.mro())  # method resolution order
'''
[<class '__main__.Bank'>, <class '__main__.CheckBalance'>, <class '__main__.Deposit'>, <class '__main__.Withdraw'>, <class 'object'>]
'''





# example1 Employee inherits from two classes → PersonalDetails and JobDetails
# Parent class 1
class PersonalDetails:
    def set_name(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)

# Parent class 2
class JobDetails:
    def set_salary(self, salary):
        self.salary = salary

    def show_salary(self):
        print("Salary:", self.salary)

# Child class (Multiple Inheritance)
class Employee(PersonalDetails, JobDetails):
    def show_details(self):
        self.show_name()
        self.show_salary()

# Create object
e = Employee()
# Set values
e.set_name('smith')
e.set_salary(50000)
# Display values
e.show_details()




# example 2 for multiple inheritance
# Parent class 1
class BankDetails:
    def set_bank(self, bank_name):
        self.bank_name = bank_name

    def show_bank(self):
        print("Bank Name:", self.bank_name)

# Parent class 2
class AccountDetails:
    def set_balance(self, balance):
        self.balance = balance

    def show_balance(self):
        print("Balance:", self.balance)

# Child class (Multiple Inheritance)
class Customer(BankDetails, AccountDetails):
    def deposit(self, amount):
        self.balance += amount
        print("After Deposit:", self.balance)

    def withdraw(self, amount):
        self.balance -= amount
        print("After Withdraw:", self.balance)

    def show_details(self):
        self.show_bank()
        self.show_balance()

# Creating object
obj = Customer()

# Setting values
obj.set_bank("SBI")
obj.set_balance(5000)

# Displaying details
obj.show_details()
obj.deposit(2000)
obj.withdraw(1000)












```
