```python
# example for encpasulation with private variable

# Example for Private Access modifier

class Bank:
    def __init__(self, name, balance):
        self.__name = name      # Private variable
        self.__balance = balance  # Private variable

    def __showdetails(self):     # Private method
        print(f"Name: {self.__name}, Balance: {self.__balance}")

class Customer(Bank):
    def display(self):
        # Trying to access private method from subclass
        try:
            self.__showdetails()
        except AttributeError as e:
            print("Error:", e)

# Creating object
obj = Customer('smith', 98000)
obj.display()

# Accessing private variable outside the class
try:
    print(obj.__balance)
except AttributeError as e:
    print("Error:", e)

# Accessing private variable outside the class using name mangling
print(obj._Bank__balance)   # Output: 98000
```

```python
class Employee:
    def __init__(self, empno, ename, salary):
        self.empno = empno        # Public variable
        self.ename = ename        # Public variable
        self.__salary = salary    # Private variable (data hiding)

    # Getter method
    def get_salary(self):
        return self.__salary

    # Setter method
    def set_salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
        else:
            print("Salary must be greater than zero")

# Example usage
e = Employee(7839, "Smith", 800)

print(f"Employee No: {e.empno}")
print(f"Employee Name: {e.ename}")

# Accessing private variable using getter
print(f"Employee Salary: {e.get_salary()}")

# Updating private variable using setter
e.set_salary(98000)
print(f"Updated Salary: {e.get_salary()}")

# Trying to set invalid salary
e.set_salary(-500)   # Will print error message
```

```python
# example for runtime polymorphism

class Parent:
    def display(self):
        print("This is Parent class")

class Child(Parent):
    def display(self):
        print("This is Child class")

# Runtime polymorphism
obj1 = Parent()
obj2 = Child()

obj1.display()   # Output: This is Parent class
obj2.display()   # Output: This is Child class
```

```python
# examplle for method overriding using with super keyword
class Parent:
    def display(self):
        print("This is Parent class")

class Child(Parent):
    def display(self):
        super().display()   # Call parent method
        print("This is Child class")

obj = Child()
obj.display()

# Output:
# This is Parent class
# This is Child class
```

```python
# example for multiple inheritance using mro 

# WAP to create same method name in multiple classes

class A:
    def display(self):
        print("class A")

class B:
    def display(self):
        print("class B")

class C(B, A):   # Multiple inheritance
    def display(self):
        print("class C")

# Object of C
obj = C()
obj.display()   # Output: class C

# Method Resolution Order (MRO)
print(C.__mro__)
# Output: (<class '__main__.C'>, <class '__main__.B'>, <class '__main__.A'>, <class 'object'>)
```

```python
# example for method overloading using  *args and **kwargs

class Calculator:
    # Method that can handle addition with variable positional arguments
    def add(self, *args):
        return sum(args)
    
    # Method that can handle operations with keyword arguments
    def calculate(self, operation, **kwargs):
        if operation == 'add':
            return sum(kwargs.values())
        elif operation == 'multiply':
            result = 1
            for value in kwargs.values():
                result *= value
            return result
        else:
            return "Unsupported operation"

# Usage
calc = Calculator()

# Using *args for positional arguments
print(calc.add(10, 20))          # Output: 30
print(calc.add(1, 2, 3, 4, 5))   # Output: 15

# Using **kwargs for keyword arguments
print(calc.calculate('add', a=10, b=20, c=30))      # Output: 60
print(calc.calculate('multiply', x=2, y=3, z=4))    # Output: 24
```

```python
# what is static method explain with an example

# Static Method: A static method is a method that belongs to the class rather than an instance of the class. It does not require an instance to be called and does not have access to instance variables (self) or class variables (cls). Static methods are defined using the @staticmethod decorator.
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b
# Usage
print(MathUtils.add(10, 20))       # Output: 30 
print(MathUtils.multiply(10, 20))  # Output: 200
```

```python
# explain class method with an example

# Class Method: A class method is a method that is bound to the class and not the instance. It can access and modify class variables but not instance variables. Class methods are defined using the @classmethod decorator and take 'cls' as the first parameter.

class Employee:
    company = "TechCorp"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

# Usage
emp = Employee("John")
print(emp.company)  # Output: TechCorp

Employee.change_company("NewCorp")
print(emp.company)  # Output: NewCorp
```

```python
# example for multi level inheritance using constructor chaining and method chaining

class A:
    def __init__(self):
        print("Constructor of A")

    def method_a(self):
        print("Method of A")
        return self  # Return self for method chaining

class B(A):
    def __init__(self):
        super().__init__()  # Constructor chaining
        print("Constructor of B")

    def method_b(self):
        print("Method of B")
        return self  # Return self for method chaining

class C(B):
    def __init__(self):
        super().__init__()  # Constructor chaining
        print("Constructor of C")

    def method_c(self):
        print("Method of C")
        return self  # Return self for method chaining

# Creating object of class C
obj = C()
# Output:
# Constructor of A
# Constructor of B
# Constructor of C

# Method chaining
obj.method_a().method_b().method_c()
# Output:
# Method of A
# Method of B
# Method of C
```

```python
# example for constructor overloading

# Python doesn't support constructor overloading directly like other languages.
# Instead, use default arguments to simulate overloading.

class Person:
    def __init__(self, name, age=18):
        self.name = name
        self.age = age

# Usage
p1 = Person("John")        # Uses default age 18
p2 = Person("Jane", 25)    # Overrides age

print(f"{p1.name} is {p1.age} years old")  # Output: John is 18 years old
print(f"{p2.name} is {p2.age} years old")  # Output: Jane is 25 years old

class Employee:
    def __init__(self, name, salary=50000):
        self.name = name
        self.salary = salary

# Usage
e1 = Employee("Smith")           # Default salary
e2 = Employee("Martin", 60000)   # Custom salary

print(f"{e1.name} earns {e1.salary}")  # Output: Smith earns 50000
print(f"{e2.name} earns {e2.salary}")  # Output: Martin earns 60000
```

```python
# example for abstract class with multiple methods
from abc import ABC, abstractmethod

# Abstract Class with multiple abstract methods
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def fuel_type(self):
        pass

# Concrete Class implementing all abstract methods
class Car(Vehicle):
    def start(self):
        print('Car starts with key')

    def stop(self):
        print('Car stops with brake')

    def fuel_type(self):
        print('Car uses petrol')

# Usage
obj = Car()
obj.start()     # Output: Car starts with key
obj.stop()      # Output: Car stops with brake
obj.fuel_type() # Output: Car uses petrol
```

```python
# wap to create online shopping cart using encapsulation

class ShoppingCart:
    def __init__(self):
        self.__items = []  # Private list of items
        self.__total = 0.0  # Private total price

    # Getter for items
    def get_items(self):
        return self.__items.copy()  # Return a copy to prevent external modification

    # Getter for total
    def get_total(self):
        return self.__total

    # Method to add item (encapsulated)
    def add_item(self, item_name, price):
        if price > 0:
            self.__items.append({"name": item_name, "price": price})
            self.__total += price
            print(f"Added {item_name} for ${price}")
        else:
            print("Invalid price")

    # Method to remove item (encapsulated)
    def remove_item(self, item_name):
        for item in self.__items:
            if item["name"] == item_name:
                self.__items.remove(item)
                self.__total -= item["price"]
                print(f"Removed {item_name}")
                return
        print(f"Item {item_name} not found")

    # Method to display cart (encapsulated)
    def display_cart(self):
        if not self.__items:
            print("Cart is empty")
        else:
            print("Items in cart:")
            for item in self.__items:
                print(f"- {item['name']}: ${item['price']}")
            print(f"Total: ${self.__total}")

# Usage
cart = ShoppingCart()
cart.add_item("Laptop", 1000)
cart.add_item("Mouse", 50)
cart.display_cart()
cart.remove_item("Mouse")
cart.display_cart()
```

```python
# example for bank system with transaction using encapsulation and file handling in simple

class BankAccount:
    def __init__(self, account_number):
        self.__account_number = account_number
        self.__balance = 0.0
        self.__transactions = []
        self.load_data()

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__transactions.append(f"Deposited ${amount}")
            self.save_data()
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            self.__transactions.append(f"Withdrew ${amount}")
            self.save_data()
            print(f"Withdrew ${amount}. New balance: ${self.__balance}")
        else:
            print("Insufficient funds or invalid amount")

    def get_balance(self):
        return self.__balance

    def get_transactions(self):
        return self.__transactions.copy()

    def save_data(self):
        with open(f"{self.__account_number}_balance.txt", "w") as f:
            f.write(str(self.__balance))
        with open(f"{self.__account_number}_transactions.txt", "w") as f:
            for t in self.__transactions:
                f.write(t + "\n")

    def load_data(self):
        try:
            with open(f"{self.__account_number}_balance.txt", "r") as f:
                self.__balance = float(f.read().strip())
            with open(f"{self.__account_number}_transactions.txt", "r") as f:
                self.__transactions = [line.strip() for line in f if line.strip()]
        except FileNotFoundError:
            pass  # No data yet

# Usage
account = BankAccount("12345")
account.deposit(1000)
account.withdraw(200)
print(f"Current balance: ${account.get_balance()}")
print("Transactions:", account.get_transactions())
```
